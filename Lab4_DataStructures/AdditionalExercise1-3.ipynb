```python
import random
import pandas as pd

LAST_NAME = "GONZALES"
SEED_NUM = 4017
FAVORITE_ARTIST = "IV of Spades"

# Set seed for reproducibility based on student data
random.seed(SEED_NUM)

print("--- 1. Generated Test Data ---")
# ---------------------------------------------------------
# Rule: Test Sources (T1-T5 using LAST_NAME)
test_sources = [f"{LAST_NAME}_{i}" for i in range(1, 6)]

# Rule: Measurement Categories
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Rule: Measurement Values (Generate using SEED_NUM and FAVORITE_ARTIST character positions)
def generate_values():
    # Convert artist name to ASCII values to use as generation basis
    artist_ascii = [ord(char) for char in FAVORITE_ARTIST if char != ' ']
    vals = []
    for _ in range(5):
        # Create numbers influenced by both SEED_NUM and the artist's name
        val = random.randint(10, 99) + (artist_ascii[0] % 10) + (SEED_NUM % 10)
        vals.append(val)
    return vals

# Build the raw dataset
records = []
# Rule: Include multiple measurements under at least two categories
# We will generate 2 measurements per category per source for all categories.
for source in test_sources:
    for cat in categories:
        for _ in range(2): # Multiple measurements
            value = generate_values()[0] # Get one generated value
            records.append({"Source": source, "Category": cat, "Value": value})

# Rule: Include one repeated test-source/category record
repeated_record = {"Source": "SMITH_1", "Category": "TEMPERATURE", "Value": 55}
records.append(repeated_record)

# Create DataFrame for easier processing
df = pd.DataFrame(records)
print(df.head(10).to_string(index=False))
print(f"...and {len(df)-10} more rows.\n")


print("--- 2 & 5. Organized and Processed Results ---")
# ---------------------------------------------------------
# Requirement: Organize so data can be analyzed by source and category
# Requirement: Arrange results in a clear/meaningful order (Sorting)
df_sorted = df.sort_values(by=["Source", "Category", "Value"]).reset_index(drop=True)
print(df_sorted.head(10).to_string(index=False))
print(f"...and {len(df_sorted)-10} more rows.\n")


print("--- 3. Repeated/Distinct Data ---")
# ---------------------------------------------------------
# Requirement: Identify repeated and distinct data
# Check for exact duplicates (same source, category, AND value)
duplicates = df_sorted[df_sorted.duplicated(keep=False)]

# To show distinct combinations of Source/Category
distinct_combos = df_sorted[['Source', 'Category']].drop_duplicates()

print(f"Total repeated rows found (including original): {len(duplicates)}")
if not duplicates.empty:
    print("Repeated Rows:")
    print(duplicates.to_string(index=False))
print(f"\nTotal distinct Source/Category combinations: {len(distinct_combos)}\n")


print("--- 4. Numerical and Categorical Summaries ---")
# ---------------------------------------------------------
# Requirement: Determine relevant numerical and categorical summaries
print("Numerical Summary (Overall):")
print(df_sorted['Value'].describe().to_string())

print("\nCategorical Summary (Count per Category):")
print(df_sorted['Category'].value_counts().to_string())
print("\n")


print("--- 6 & 7. Final Consolidated Summary ---")
# ---------------------------------------------------------
# Requirement: Present original/processed info in structured form
# Requirement: Produce a consolidated summary
# Group by Source and Category to get stats
summary_df = df_sorted.groupby(['Source', 'Category']).agg(
    Total_Measurements=('Value', 'count'),
    Average_Value=('Value', 'mean'),
    Min_Value=('Value', 'min'),
    Max_Value=('Value', 'max')
).round(2).reset_index()

print("FINAL ORGANIZED AND SUMMARIZED TEST DATASET:")
print(summary_df.to_string(index=False))
```
