```python
import random
import pandas as pd

LAST_NAME = "Gonzales"
SEED_NUM = 4017
FAVORITE_ARTIST = "IV of Spades"

# Set seed for reproducibility based on student data
random.seed(SEED_NUM)

print("--- 1. Generated Test Data ---")
# ---------------------------------------------------------
# Rule: Test Sources (T1-T5 using LAST_NAME)
test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

# Rule: Measurement Categories
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# Rule: Measurement Values (Generate using SEED_NUM and FAVORITE_ARTIST character positions)
def generate_value():
    # Convert artist name to ASCII values to use as generation basis
    artist_ascii = [ord(char) for char in FAVORITE_ARTIST if char != ' ']
    
    # Create a number influenced by both SEED_NUM and the artist's name
    base_val = random.randint(10, 99)
    artist_mod = artist_ascii[0] % 10
    seed_mod = SEED_NUM % 10
    return base_val + artist_mod + seed_mod

# Build the raw dataset
records = []
# Rule: Include multiple measurements under at least two categories
# We will generate 2 measurements per category per source.
for source in test_sources:
    for cat in categories:
        for _ in range(2): 
            value = generate_value()
            records.append({"Source": source, "Category": cat, "Value": value})

# Rule: Include one repeated test-source/category record.
# Grab a record that already exists and append it again to create an exact duplicate.
repeated_record = {"Source": "Gonzales_T1", "Category": "TEMPERATURE", "Value": records[0]['Value']}
records.append(repeated_record)

# Create DataFrame for easier processing
df = pd.DataFrame(records)
print(df.head(10).to_string(index=False))
print(f"...and {len(df)-10} more rows.\n")


print("--- 2 & 5. Organized and Processed Results ---")
# ---------------------------------------------------------
# Requirement: Organize so data can be analyzed by source and category
# Requirement: Arrange results in a clear/meaningful order (Sorting)
df_sorted = df.sort_values(by=["Source", "Category", "Value"]).reset_index(drop=True)
print(df_sorted.head(10).to_string(index=False))
print(f"...and {len(df_sorted)-10} more rows.\n")


print("--- 3. Repeated/Distinct Data ---")
# ---------------------------------------------------------
# Requirement: Identify repeated and distinct data
# Check for exact duplicates (same source, category, AND value)
duplicates = df_sorted[df_sorted.duplicated(keep=False)]

# To show distinct combinations of Source/Category
distinct_combos = df_sorted[['Source', 'Category']].drop_duplicates()

print(f"Total repeated rows found (including original): {len(duplicates)}")
if not duplicates.empty:
    print("Repeated Rows:")
    print(duplicates.to_string(index=False))
print(f"\nTotal distinct Source/Category combinations: {len(distinct_combos)}\n")


print("--- 4. Numerical and Categorical Summaries ---")
# ---------------------------------------------------------
# Requirement: Determine relevant numerical and categorical summaries
print("Numerical Summary (Overall):")
print(df_sorted['Value'].describe().to_string())

print("\nCategorical Summary (Count per Category):")
print(df_sorted['Category'].value_counts().to_string())
print("\n")


print("--- 6 & 7. Final Consolidated Summary ---")
# Requirement: Present original/processed info in structured form
# Requirement: Produce a consolidated summary
# Group by Source and Category to get stats
summary_df = df_sorted.groupby(['Source', 'Category']).agg(
    Total_Measurements=('Value', 'count'),
    Average_Value=('Value', 'mean'),
    Min_Value=('Value', 'min'),
    Max_Value=('Value', 'max')
).round(2).reset_index()

print("FINAL ORGANIZED AND SUMMARIZED TEST DATASET:")
print(summary_df.to_string(index=False))
```

```python
import random
import pandas as pd
import numpy as np

LAST_NAME = "Gonzales"
SEED_NUM = 4017
FAVORITE_ARTIST = "IV of Spades"

# Set seed for reproducibility
random.seed(SEED_NUM)
np.random.seed(SEED_NUM)

print("--- 1. Generated Performance Records ---")
# ---------------------------------------------------------
# Rule: Student Names (Combine LAST_NAME with 5 labels)
student_names = [f"{LAST_NAME}_{i}" for i in range(1, 6)]

# Rule: Subjects
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

# Rule: Scores (using SEED_NUM and FAVORITE_ARTIST character positions)
def generate_score():
    # Base random score between 50 and 100
    base_score = random.randint(50, 100)
    
    # Add influence from artist name (sum of ASCII values of first 3 letters mod 10)
    artist_val = sum(ord(c) for c in FAVORITE_ARTIST[:3] if c != ' ') % 10
    
    # Add influence from SEED_NUM (last digit)
    seed_val = SEED_NUM % 10
    
    # Combine and ensure it doesn't exceed 100
    final_score = base_score + artist_val + seed_val
    return min(final_score, 100) 

# Rule: Performance Category
def get_category(score):
    if pd.isna(score):
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 75:
        return "GOOD"
    elif score >= 60:
        return "SATISFACTORY"
    else:
        return "NEEDS IMPROVEMENT"

# Build the raw dataset
records = []
for student in student_names:
    for subject in subjects:
        score = generate_score()
        records.append({
            "Student": student, 
            "Subject": subject, 
            "Score": score
        })

# Rule: Include one repeated student-subject record
# We will duplicate the very first record generated
repeated_record = records[0].copy()
records.append(repeated_record)

# Rule: Include one student record with a missing score
incomplete_record = {"Student": "Gonzales_5", "Subject": "MATHEMATICS", "Score": None}
records.append(incomplete_record)

# Create DataFrame
df = pd.DataFrame(records)
print(df.to_string(index=False))
print("\n")


print("--- 2, 4 & 6. Organized Data, Repeated/Incomplete Records ---")
# ---------------------------------------------------------
# Requirement 2: Organize so student and subject performance can be analyzed together
# Requirement 6: Arrange results in a clear order
df_sorted = df.sort_values(by=["Student", "Subject", "Score"], na_position="last").reset_index(drop=True)

# Requirement 4: Identify repeated and incomplete records
# Find exact duplicates (same student, subject, AND score)
duplicates = df_sorted[df_sorted.duplicated(keep=False) & df_sorted['Score'].notna()]
# Find incomplete records
incomplete = df_sorted[df_sorted['Score'].isna()]

print("Identified Repeated Record(s):")
print(duplicates.to_string(index=False))
print("\nIdentified Incomplete Record(s):")
print(incomplete.to_string(index=False))
print("\n")


print("--- 3. Determine Performance Categories ---")
# ---------------------------------------------------------
# Requirement 3: Determine the performance category of each valid score
df_sorted['Performance_Category'] = df_sorted['Score'].apply(get_category)
print(df_sorted[['Student', 'Subject', 'Score', 'Performance_Category']].head(10).to_string(index=False))
print("...\n")


print("--- 5. Student and Subject Summaries ---")
# ---------------------------------------------------------
# Requirement 5: Determine relevant performance summaries for students and subjects
# We exclude NaN values automatically when calculating mean and count
student_summary = df_sorted.groupby('Student').agg(
    Total_Valid_Scores=('Score', 'count'),
    Average_Score=('Score', 'mean'),
    Highest_Score=('Score', 'max'),
    Lowest_Score=('Score', 'min')
).round(2)

subject_summary = df_sorted.groupby('Subject').agg(
    Total_Valid_Scores=('Score', 'count'),
    Average_Score=('Score', 'mean')
).round(2)

print("Student Summaries:")
print(student_summary.to_string())
print("\nSubject Summaries:")
print(subject_summary.to_string())
print("\n")


print("--- 7 & FINAL OUTPUT. Consolidated Student Performance Report ---")
# ---------------------------------------------------------
# Requirement 7: Present a consolidated report on the student performance data.
# We will create a pivot table to show Subjects as columns, Students as rows.

# Drop the exact duplicate record so it doesn't skew the pivot average
df_clean = df_sorted.drop_duplicates(subset=['Student', 'Subject', 'Score'])

final_report = pd.pivot_table(
    df_clean, 
    values='Score', 
    index='Student', 
    columns='Subject', 
    aggfunc='mean'
).round(1)

# Fill NaN for the missing score with a dash for readability in the final report
final_report = final_report.fillna('-')

print("FINAL STUDENT PERFORMANCE REPORT (Average Scores by Subject):")
print(final_report.to_string())
```

```python
import random
import pandas as pd

LAST_NAME = "Gonzales"
SEED_NUM = 4017
FAVORITE_ARTIST = "IV of Spades"

# Set seed for reproducibility based on student data
random.seed(SEED_NUM)

print("--- 1. Generated Transaction Records ---")
# ---------------------------------------------------------
# Rule: Customers (Combine LAST_NAME with C1-C5)
customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]

# Rule: Products and Fixed Prices
product_prices = {
    "NOTEBOOK": 50.00,
    "PEN": 15.00,
    "USB": 250.00,
    "HEADPHONES": 1200.00,
    "CALCULATOR": 850.00
}
products = list(product_prices.keys())

# Rule: Quantities (Generate using SEED_NUM and FAVORITE_ARTIST character positions)
def generate_quantity():
    # Get ASCII values of first 3 letters of artist name
    artist_ascii = [ord(c) for c in FAVORITE_ARTIST[:3] if c != ' ']
    artist_sum = sum(artist_ascii)
    
    # Base random quantity between 1 and 5
    base_qty = random.randint(1, 5)
    
    # Combine using seed num and artist values
    calculated_qty = base_qty + (SEED_NUM % 3) + (artist_sum % 2)
    return calculated_qty

# Rule: Generate multiple transactions
records = []
for customer in customers:
    # Give each customer 2 to 3 random transactions
    num_transactions = random.randint(2, 3)
    for _ in range(num_transactions):
        product = random.choice(products)
        qty = generate_quantity()
        records.append({
            "Customer": customer,
            "Product": product,
            "Quantity": qty
        })

# Rule: Include at least one repeated customer-product transaction
# We will duplicate the first generated record
repeated_record = records[0].copy()
records.append(repeated_record)

# Create DataFrame
df = pd.DataFrame(records)

# Requirement 3: Determine the total transaction value for each record
df['Total_Value'] = df.apply(lambda row: row['Quantity'] * product_prices[row['Product']], axis=1)

print(df.to_string(index=False))
print(f"\nTotal records generated: {len(df)}\n")


print("--- 2, 4 & 6. Organized Data, Repeated/Distinct, Transaction Values ---")
# ---------------------------------------------------------
# Requirement 2: Organize so customer, product, qty, and value can be analyzed together
# Requirement 6: Arrange results in a clear and meaningful order
df_sorted = df.sort_values(by=["Customer", "Product", "Quantity"]).reset_index(drop=True)

# Requirement 4: Identify repeated purchases and distinct products or customers
# Repeated purchases (exact duplicates in customer, product, and quantity)
duplicates = df_sorted[df_sorted.duplicated(keep=False)]
distinct_customers = df_sorted['Customer'].nunique()
distinct_products = df_sorted['Product'].nunique()

print("Identified Repeated Purchase(s):")
if duplicates.empty:
    print("None found.")
else:
    print(duplicates.to_string(index=False))
print(f"\nTotal Distinct Customers: {distinct_customers}")
print(f"Total Distinct Products: {distinct_products}\n")

print("Organized Transaction Data (Showing first 10 rows):")
print(df_sorted.head(10).to_string(index=False))
print("...\n")


print("--- 5. Customer and Product Summaries ---")
# ---------------------------------------------------------
# Requirement 5: Determine relevant summaries by customer and product

# Customer Summary
customer_summary = df_sorted.groupby('Customer').agg(
    Total_Transactions=('Product', 'count'),
    Total_Items_Bought=('Quantity', 'sum'),
    Total_Spent=('Total_Value', 'sum')
).round(2)

# Product Summary
product_summary = df_sorted.groupby('Product').agg(
    Times_Purchased=('Customer', 'count'),
    Total_Quantity_Sold=('Quantity', 'sum'),
    Total_Revenue=('Total_Value', 'sum')
).round(2)

print("CUSTOMER SUMMARIES:")
print(customer_summary.to_string())
print("\nPRODUCT SUMMARIES:")
print(product_summary.to_string())
print("\n")


print("--- 7 & 8. Consolidated Transaction Report ---")
# ---------------------------------------------------------
# Requirement 7 & 8: Present a consolidated report showing major transaction results.
# We will drop the exact duplicate to not skew the consolidated totals, then create a pivot table.

df_clean = df_sorted.drop_duplicates(subset=['Customer', 'Product', 'Quantity'])

final_report = pd.pivot_table(
    df_clean,
    values='Total_Value',
    index='Customer',
    columns='Product',
    aggfunc='sum',
    fill_value=0
).round(2)

# Calculate Overall Total
overall_total = df_clean['Total_Value'].sum()

print("FINAL CONSOLIDATED TRANSACTION REPORT (Total Spent per Customer per Product):")
print(final_report.to_string())
print(f"\nOVERALL TOTAL REVENUE: PHP {overall_total:,.2f}")
```
